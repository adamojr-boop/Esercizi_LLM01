```python

# TODO Inizializza il client di Open AI

import os, openai
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env') # Read the .env file

# Configure OpenAI client
openai.api_key = os.environ['OPENAI_API_KEY']

client = OpenAI()
```

```python
input_text = input("Enter text: ")


complention = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "sei un assistente in grado di analizzare testi e creare domande educative basate sul contenuto che ti viene passato. Sei in grado di generare domande chiare e pertinenti per scopi educativi(es. a risposta multipla, a risposta aperta, o un mix)."},
        {"role": "user", "content": input_text}
    ],
)

print(complention.choices[0].message.content)
```
