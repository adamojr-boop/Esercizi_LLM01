**Obiettivo:** Sviluppare un prompt che sia adatto alla traduzione fluida tra italiano e inglese.

```python

# TODO Inizializza il client di Open AI
import os, openai
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env') # Read the .env file

# Configure OpenAI client
openai.api_key = os.environ['OPENAI_API_KEY']

client = OpenAI()
```

```python
input_text = input("Enter text: ")

complention = client.chat.completions.create(
    model="gpt-3.5-turbo",
    messages=[
        {"role": "system", "content": "Sei un traduttore professionista. Traduci il testo inserito dall'utente: se è in italiano, traducilo in inglese; se è in inglese, traducilo in italiano. Mantieni il testo fluido e naturale."},
        {"role": "user", "content": input_text}
    ],
)

print(complention.choices[0].message.content)
```
