```python

# TODO Inizializza il client di Open AI

import os, openai
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env') # Read the .env file

# Configure OpenAI client
openai.api_key = os.environ['OPENAI_API_KEY']

client = OpenAI()
```

```python
data = {"product": "Zaino da trekking", "price": 89.99, "material": "Impermeabile", "colors": ["nero", "verde scuro"], "stock": 14}


complention = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "sei un assistente in grado di analizzare un set di dati strutturali e fornire sintesi con descrizioni testuali chiare e concise"},
        {"role": "user", "content": str(data)}
    ],
)

print(complention.choices[0].message.content)
```
